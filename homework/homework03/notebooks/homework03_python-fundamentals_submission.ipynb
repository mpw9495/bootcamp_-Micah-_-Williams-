## 1. NumPy Operations

```python
# TODO: Create a NumPy array and perform vectorized operations
arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("Array multiplied by 2:", arr * 2)
print("Array squared:", arr ** 2)
print("Array plus 10:", arr + 10)
```

## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit

arr = np.arange(10000)

def loop_square(arr):
    result = []
    for x in arr:
        result.append(x ** 2)
    return result

%timeit loop_square(arr)
%timeit arr ** 2
```

## 3. Load and Inspect Dataset

```python
# TODO: Load CSV and inspect with .head(), .info(), .describe()

from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "starter_data.csv"

df = pd.read_csv(DATA_PATH)

print("First 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nSummary statistics:")
display(df.describe())
```

## 4. Summary Statistics and Groupby

```python
# TODO: Perform groupby and save output to data/processed/

summary = df.groupby("category")["value"].agg(["count", "mean", "min", "max"])

print("Summary by category:")
display(summary)

OUTPUT_PATH = PROJECT_ROOT / "data" / "processed" / "summary.csv"

summary.to_csv(OUTPUT_PATH)

print("Saved summary to:", OUTPUT_PATH)
```

## 5. Reusable Function

```python
def calculate_summary(data):
    """
    Calculate summary statistics for the value column.
    """
    return data["value"].describe()


summary_stats = calculate_summary(df)

print("Summary statistics using reusable function:")
display(summary_stats)
```
